```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 10, 100)
y=x**2
fig, ax = plt.subplots()
ax.plot (x,y)
ax.set_xlabel("this is my x")
ax.set_ylabel("y axis")
plt.show()
```

```python
x=np.linspace(-2,2)
y1=x**2
y2=x**3
```

```python
plt.plot(x, y1, label='y1 = x^2')
plt.plot(x, y2, label= 'y2 = x^3')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Plot of y = x^2 and y = x^3')
plt.grid(True)

plt.show()
```
