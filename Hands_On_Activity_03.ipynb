```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
integrins=pd.read_excel('gtex_integrin_7_organs (2).xlsx')
```

```python
integrins
```

```python
#pd.set_option('display.max_rows', None)     #shows all rows, no set maximum to the number of rows displayed
#pd.set_option('display.max_columns', None)     #shows all rows, no set maximum to the number of rows displayed
#pd.reset_option('display.max_rows')      #back to default settings for rows displayed
#pd.reset_option('display.max_columns')      #back to default settings for columns displayed
breast_integrins = integrins[integrins['primary_site'] == 'Breast']
breast_integrins
```

```python
#violin plot for all the genes of the breast
plt.figure(figsize = (16, 6))
sns.violinplot(data = breast_integrins)
plt.title("Integrin Genes of the Breast")
plt.xlabel("Integrin Genes")
plt.ylabel("Gene Expression Levels")
plt.show()
```

```python
lung_integrins = integrins[integrins['primary_site'] == 'Lung']
lung_integrins
```

```python
#violin plot for all the genes of the lung
plt.figure(figsize = (16, 6))
sns.violinplot(data = lung_integrins)
plt.title("Integrin Genes of the Lung")
plt.xlabel("Integrin Genes")
plt.ylabel("Gene Expression Levels")
plt.show()
```

```python
breast_lung_integrins = integrins[integrins['primary_site'].isin(['Breast', 'Lung'])]     #filter data by organ, display both breast and lung data
breast_lung_integrins
```

```python
df=breast_lung_integrins.drop(breast_lung_integrins.columns[0], axis=1)
df
```

```python
breast_liver_integrins = integrins[integrins['primary_site'].isin(['Breast', 'Lung'])]     #filter data by organ, display both breast and lung data

#rearrange data
breast_lung_integrins_vertical = breast_lung_integrins.melt(id_vars = 'primary_site', var_name = 'integrin_gene', value_name = 'expression_levels')
breast_lung_integrins_vertical
```

```python
breast_lung_integrins_vertical = breast_lung_integrins_vertical[
breast_lung_integrins_vertical['integrin_gene'] != 'Unnamed: 0'
]
```

```python
breast_lung_integrins_vertical['integrin_gene'].head()
breast_lung_integrins_vertical['integrin_gene'].dtype
```

```python
plt.figure(figsize=(16, 6))
sns.violinplot(x = 'integrin_gene', y = 'expression_levels', hue = 'primary_site', data = breast_lung_integrins_vertical, split = True, inner = 'quartile')
plt.title("Integrin Genes of the Breast vs. the Lung")
plt.xlabel("Integrin Gene")
plt.ylabel("Gene Expression Levels")
plt.legend(title = 'primary_site')
plt.show()
```

```python
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler

integrins
```

```python
X = integrins.drop(columns=["primary_site", "Unnamed: 0"], errors="ignore")
labels = integrins["primary_site"]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

tsne = TSNE(n_components=2, random_state=42)
X_2d = tsne.fit_transform(X_scaled)
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    x=X_2d[:, 0],
    y=X_2d[:, 1],
    hue=labels,
    palette="tab10",
    s=30,
)

plt.title("Integrin Gene Expression Levels (t-SNE)")
plt.xlabel("t-SNE 1")
plt.ylabel("t-SNE 2")
plt.legend(title="primary site", bbox_to_anchor=(1.05,1), loc="upper left")
plt.tight_layout()
plt.show()
```
